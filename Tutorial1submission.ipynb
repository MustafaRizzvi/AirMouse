```python
import pandas as pd
import numpy as np
```

1. create a numpy array of size 2x2

```python
arr1 = np.array([[1,2],[3,4]])
print(arr1)
```

2. Create a random numpy array of size 3x4

```python
arr2 = np.random.rand(3,4)
print(arr2)
```

3. Create a numpy array of ones and a numpy array of zeros

```python
ones = np.ones((4,3))
zeros = np.zeros((3,4))
print(ones)
print(zeros)
```

4. Create a numpy array of size 6x3 and transpose it

```python
arr3 = np.random.rand(6,3)
print(arr3)
print(arr3.T)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n, p, m = 3, 4, 2

arr_np = np.random.rand(n, p)
arr_pm = np.random.rand(p, m)

dot_product = np.dot(arr_np, arr_pm)
print(arr_np)
print(arr_pm)
print("\nDot product (size n x m):\n", dot_product)
```
